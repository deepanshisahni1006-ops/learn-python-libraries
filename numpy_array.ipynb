ARRAY CREATION

```python
import numpy as np
# np.array
a=np.array([1,2,3])
print(a)
```

```python
# 2D and 3D
b=np.array([[1,2,3],[4,5,6]])
print(b)
c=np.array([[1,2],[3,4],[5,6]])
print(c)
```

```python
d=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(d)
```

```python
# dtype
np.array([1,2,3],dtype=float)
```

```python
# np.arange
np.arange(1,11,2)
```

```python
# with reshape
np.arange(1,13).reshape(4,3)
```

```python
# np.zeros and np.ones
np.zeros((3,4))
```

```python
np.ones((4,4))
```

```python
# np.random
np.random.random((2,5))
```

```python
# np.linspace
np.linspace(-10,10,10)
```

```python
# np.identity
np.identity(4)
```

ARRAY ATTRIBUTES

```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)

a3
```

```python
# ndim
a3.ndim
```

```python
# shape
a3.shape
```

```python
# size
a2.size
```

```python
# itemsize
a2.itemsize
```

```python
# dtype
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
# changing datatype
# astype
a3.astype(np.int32)
```

ARRAY OPERATIONS

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)

a2
```

```python
# scalar operations
# arithmetic
a1**2
```

```python
# relational
a1>5
```

```python
# vector operations
# arithmetic
a1*a2
```

ARRAY FUNCTIONS

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
# max,min,sum,prod
# 0 - column and 1 - row
np.sum(a1,axis=0)
```

```python
# mean,median,std,var
# can use axis
np.mean(a1)
```

```python
# trignometric functions
np.sin(a1)
```

```python
# dot product
a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)

np.dot(a2,a3)
```

```python
# log and exponents
np.exp(a1)
```

```python
# round,floor,ceil
np.round(np.random.random((2,3))*100)
```
