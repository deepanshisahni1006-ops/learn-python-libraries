```python
import numpy as np
# np.array
a=np.array([1,2,3])
print(a)
```

```python
# 2D and 3D
b=np.array([[1,2,3],[4,5,6]])
print(b)
c=np.array([[1,2],[3,4],[5,6]])
print(c)
```

```python
d=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(d)
```

```python
# dtype
np.array([1,2,3],dtype=float)
```

```python
# np.arange
np.arange(1,11,2)
```

```python
# with reshape
np.arange(1,13).reshape(4,3)
```

```python
# np.zeros and np.ones
np.zeros((3,4))
```

```python
np.ones((4,4))
```

```python
# np.random
np.random.random((2,5))
```

```python
# np.linspace
np.linspace(-10,10,10)
```

```python
# np.identity
np.identity(4)
```
