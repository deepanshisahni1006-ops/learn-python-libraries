```python
import numpy as np
import pandas as pd
```

SERIES FROM LISTS

```python
# string
country=['India','Pakistan','Nepal','Srilanka']

pd.Series(country)
```

```python
# integers
runs=[54,36,68,45,86,34]
pd.Series(runs)
```

```python
# custom index
country=['India','Pakistan','Nepal','Srilanka']
pd.Series(country,index=['a','b','c','d'])
```

```python
marks=[65,98,78,54]
subjects=['English','Maths','Hindi','Science']
pd.Series(marks,index=subjects)
```

```python
# setting a name
pd.Series(marks,index=subjects,name='Nitish ke marks')
```

SERIES FROM DICTIONARIES

```python
marks={
    'maths': 98,
    'english':54,
    'science':58,
    'hindi':78
}
pd.Series(marks,name='Nitish ke marks')
```

SERIES ATTRIBUTES

```python
# size
marks={
    'maths': 98,
    'english':54,
    'science':58,
    'hindi':78
}
marks_series=pd.Series(marks,name='Nitish ke marks')
marks_series.size
```

```python
# dtype
marks_series.dtype
```

```python
# name
marks_series.name
```

```python
# is_unique
marks_series.is_unique
```

```python
pd.Series([1,1,2,3,4,5,6]).is_unique
```

```python
# index
marks_series.index
```

```python
# values
marks_series.values
```
